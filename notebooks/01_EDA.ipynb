## 5. Tổng kết EDA

### Phát hiện chính:
1. **German Credit**: Dữ liệu sạch, ít giá trị thiếu. Mất cân bằng lớp vừa phải (~70:30).
2. **Give Me Some Credit**: Có giá trị thiếu ở MonthlyIncome và NumberOfDependents. Mất cân bằng nghiêm trọng (~93:7).
3. Cả 2 bộ dữ liệu đều có ngoại lai cần xử lý.
4. Một số đặc trưng có tương quan cao → cần xem xét khi chọn đặc trưng.

### Bước tiếp theo:
- Xử lý giá trị thiếu và ngoại lai
- Mã hóa biến phân loại
- Xử lý mất cân bằng lớp (SMOTE)
- Trích xuất đặc trưng mới
